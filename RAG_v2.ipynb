```python
import os
import pandas as pd

from langchain_community.document_loaders import (
    PyPDFLoader,
    Docx2txtLoader,
    TextLoader
)
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

folder = "/content/RAG/data"
documents = []

# 1. Load PDF
pdf_path = os.path.join(folder, "leave_policy_plan.pdf")
documents.extend(PyPDFLoader(pdf_path).load())

# 2. Load Word document
docx_path = os.path.join(
    folder, "project_management_activities_corporate.docx"
)
documents.extend(Docx2txtLoader(docx_path).load())

# 3. Load text file
txt_path = os.path.join(folder, "remote_work_policy.txt")
documents.extend(TextLoader(txt_path, encoding="utf-8").load())

# 4. Load Excel file
excel_path = os.path.join(
    folder, "posh_applicability_checker_expanded.xlsx"
)
sheets = pd.read_excel(excel_path, sheet_name=None)

for sheet_name, df in sheets.items():
    documents.append(
        Document(
            page_content=df.to_string(index=False),
            metadata={
                "source": "posh_applicability_checker_expanded.xlsx",
                "sheet": sheet_name
            }
        )
    )

print("Documents/pages loaded:", len(documents))

# Split documents into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,
    separators=["\n\n", "\n", " ", ""]
)

chunks = splitter.split_documents(documents)

print("Total chunks created:", len(chunks))

# Display the chunks
for i, chunk in enumerate(chunks):
    print("\n" + "=" * 60)
    print("CHUNK:", i + 1)
    print("SOURCE:", chunk.metadata.get("source", "N/A"))
    print("PAGE:", chunk.metadata.get("page", "N/A"))
    print("SHEET:", chunk.metadata.get("sheet", "N/A"))
    print("CHARACTERS:", len(chunk.page_content))
    print("-" * 60)
    print(chunk.page_content)

print("\nRecursive chunking completed!")
```

convert the chunks to vector numerical embeddings

```python
from sentence_transformers import SentenceTransformer

# Load the MiniLM embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Extract the text from all 123 chunks
chunk_texts = [chunk.page_content for chunk in chunks]

# Convert every chunk into a vector embedding
embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Total chunks:", len(chunk_texts))
print("Embedding shape:", embeddings.shape)
print("\nFirst chunk:")
print(chunk_texts[0])
print("\nFirst embedding:")
print(embeddings[0])
```

```python
import os
import pandas as pd

from langchain_community.document_loaders import (
    PyPDFLoader, Docx2txtLoader, TextLoader
)
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

folder = "/content/RAG/data"
documents = []

documents.extend(
    PyPDFLoader(os.path.join(folder, "leave_policy_plan.pdf")).load()
)

documents.extend(
    Docx2txtLoader(
        os.path.join(folder, "project_management_activities_corporate.docx")
    ).load()
)

documents.extend(
    TextLoader(
        os.path.join(folder, "remote_work_policy.txt"),
        encoding="utf-8"
    ).load()
)

sheets = pd.read_excel(
    os.path.join(folder, "posh_applicability_checker_expanded.xlsx"),
    sheet_name=None
)

for sheet_name, df in sheets.items():
    documents.append(
        Document(
            page_content=df.to_string(index=False),
            metadata={
                "source": "posh_applicability_checker_expanded.xlsx",
                "sheet": sheet_name
            }
        )
    )

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,
    separators=["\n\n", "\n", " ", ""]
)

chunks = splitter.split_documents(documents)

print("Documents/pages loaded:", len(documents))
print("Total chunks created:", len(chunks))
```

now for hte chunks create the embeddings again

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_texts = [chunk.page_content for chunk in chunks]

embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Total chunks:", len(chunk_texts))
print("Embedding shape:", embeddings.shape)
```

store embeddings in chromadb


```python
import chromadb

# Create a persistent ChromaDB database
client = chromadb.PersistentClient(
    path="/content/RAG/chroma_db"
)

# Create a collection for our documents
collection = client.get_or_create_collection(
    name="company_documents"
)

# Prepare chunk IDs and text
ids = [f"chunk_{i}" for i in range(len(chunks))]
texts = [chunk.page_content for chunk in chunks]

# Convert embeddings to ordinary Python lists
vectors = embeddings.tolist()

# Preserve source information
metadatas = []

for chunk in chunks:
    metadata = {}
    for key, value in chunk.metadata.items():
        if value is not None:
            metadata[key] = str(value)
    metadatas.append(metadata)

# Save everything to ChromaDB
collection.upsert(
    ids=ids,
    documents=texts,
    embeddings=vectors,
    metadatas=metadatas
)

print("Chunks stored:", collection.count())
print("ChromaDB storage completed!")
```

set threshold values for relevancy if closest means more relevant


```python
def search_documents(question, n_results=3, max_distance=1.0):
    query_embedding = model.encode(question).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=n_results
    )

    found_relevant = False

    for i, text in enumerate(results["documents"][0]):
        distance = results["distances"][0][i]

        if distance <= max_distance and text.strip():
            print(f"\nResult {i + 1} | Distance: {distance:.4f}")
            print("Source:", results["metadatas"][0][i].get("source", "Unknown"))
            print(text[:500])
            found_relevant = True

    if not found_relevant:
        print("No sufficiently relevant document found.")
        print("Please ask a question related to the available documents.")


search_documents("How many days of sick leave are suggested?")
search_documents("What is the capital of Japan?")
```

clean the excel


```python
import pandas as pd

excel_path = "/content/RAG/data/posh_applicability_checker_expanded.xlsx"

sheets = pd.read_excel(excel_path, sheet_name=None)

posh_documents = []

for sheet_name, df in sheets.items():

    df = df.fillna("")
    df = df.loc[
        ~df.astype(str).apply(
            lambda row: all(value.strip() == "" for value in row),
            axis=1
        )
    ]

    for _, row in df.iterrows():

        parts = []

        for column, value in row.items():
            value = str(value).strip()

            if value and not value.startswith("Unnamed:"):
                parts.append(f"{column}: {value}")

        if parts:
            text = f"Sheet: {sheet_name}\n" + "\n".join(parts)
            posh_documents.append(text)

print("Clean POSH records:", len(posh_documents))
print("\nSample record:\n")
print(posh_documents[0])
```

replace old excel chunks

```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Convert the cleaned POSH records into Documents
posh_docs = []

for i, text in enumerate(posh_documents):
    posh_docs.append(
        Document(
            page_content=text,
            metadata={
                "source": "posh_applicability_checker_expanded.xlsx",
                "record": str(i)
            }
        )
    )

# Keep the existing PDF, DOCX and TXT documents.
# Remove the old Excel documents.
other_documents = [
    doc for doc in documents
    if doc.metadata.get("source") !=
    "posh_applicability_checker_expanded.xlsx"
]

# Combine the other documents with the cleaned POSH records
documents = other_documents + posh_docs

# Split everything into chunks again
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,
    separators=["\n\n", "\n", " ", ""]
)

chunks = splitter.split_documents(documents)

print("Documents after cleaning:", len(documents))
print("Total chunks:", len(chunks))
print("\nSample POSH chunk:")
print(next(
    chunk.page_content for chunk in chunks
    if "POSH Quick Check" in chunk.page_content
))
```

```python
def retrieve_context(question, n_results=5, max_distance=1.0):
    query_embedding = model.encode(question).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=n_results
    )

    context_parts = []

    for text, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        if distance <= max_distance and text.strip():
            source = metadata.get("source", "Unknown")

            context_parts.append(
                f"Source: {source}\n{text}"
            )

    if not context_parts:
        return None

    return "\n\n---\n\n".join(context_parts)


question = "How should a POSH complaint be reported to the Internal Committee or HR?"

context = retrieve_context(question)

if context:
    print("Retrieved context:\n")
    print(context)
else:
    print("No sufficiently relevant context found.")
```

```python
def ask_rag(question):
    # 1. Retrieve relevant document passages
    context = retrieve_context(question)

    # 2. Reject questions when no relevant context is found
    if context is None:
        return "I couldn't find sufficiently relevant information in the provided documents."

    # 3. Ask Gemini to answer using the retrieved context
    prompt = f"""
You are a company document assistant.

Answer the user's question using ONLY the context below.
Do not invent facts or use outside knowledge.
If the context does not contain the answer, say so.
Keep the answer simple and clear.
Mention the source document when possible.

CONTEXT:
{context}

QUESTION:
{question}

ANSWER:
"""

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    return response.text


# Test your complete RAG pipeline
answer = ask_rag(
    "How should a POSH complaint be reported to the Internal Committee or HR?"
)

print("Answer:", answer)
```

```python
chat_history = []

def ask_rag(question):
    global chat_history

    # Retrieve relevant information from your documents
    context = retrieve_context(question)

    if context is None:
        context = "No sufficiently relevant document passages were retrieved."

    # Keep recent conversation history
    previous_chat = ""
    for message in chat_history[-6:]:
        previous_chat += message + "\n"

    prompt = f"""
You are a friendly, conversational company-document assistant.

YOUR BEHAVIOUR:
1. Respond naturally to greetings like hi, hello, and good morning.
2. If asked what you can do, explain that you can answer questions
   about the uploaded company documents.
3. If asked what you cannot answer, explain your limitations clearly.
4. Answer document-related questions using the supplied context.
5. Never invent company policies, rules, or personal employee details.
6. If the documents do not contain the answer, say so clearly.
7. For general factual questions outside the documents, explain
   that your document knowledge does not cover that topic.
8. If the user asks a follow-up, use the conversation history
   to understand what they mean.
9. Be friendly, clear, and concise.

RETRIEVED DOCUMENT CONTEXT:
{context}

PREVIOUS CONVERSATION:
{previous_chat}

CURRENT USER QUESTION:
{question}

Respond naturally and accurately.
"""

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    answer = response.text

    # Save this exchange for follow-up questions
    chat_history.append("User: " + question)
    chat_history.append("Assistant: " + answer)

    return answer
```
